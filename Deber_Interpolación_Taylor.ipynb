```python
import numpy as np
import matplotlib.pyplot as plt


def taylor_sin(x, n):
    x = np.asarray(x, dtype=float)
    termino = x.copy()
    suma = termino.copy()
    
    num_pasos = (n - 1) // 2
    for k in range(num_pasos):
        termino = termino * (-x**2) / ((2 * k + 2) * (2 * k + 3))
        suma = suma + termino
        
    return suma

x1 = np.linspace(-2 * np.pi, 2 * np.pi, 2001)
grados_g1 = [1, 3, 5, 7, 9]

plt.figure(figsize=(9, 5.5))
plt.plot(x1, np.sin(x1), 'k-', linewidth=2.5, label=r'$\sin(x)$')

estilos = ['--', '-.', ':', '--', '-.']
for n, estilo in zip(grados_g1, estilos):
    plt.plot(x1, taylor_sin(x1, n), linestyle=estilo, linewidth=1.8, label=f'$P_{{{n}}}(x)$')

plt.ylim(-2, 2)
plt.xlim(-2 * np.pi, 2 * np.pi)
plt.axhline(0, color='gray', linewidth=0.8, alpha=0.5)
plt.axvline(0, color='gray', linewidth=0.8, alpha=0.5)
plt.title(r'Aproximación de $\sin(x)$ con Polinomios de Maclaurin en $[-2\pi, 2\pi]$ Alejandro López', fontsize=12)
plt.xlabel('$x$', fontsize=11)
plt.ylabel('$y$', fontsize=11)
plt.legend(loc='upper right', frameon=True)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('grafica_1_taylor_sin.png', dpi=300)
plt.show()

grados_impares = np.arange(1, 22, 2) 

x_pi = np.linspace(-np.pi, np.pi, 2001)
x_pi4 = np.linspace(-np.pi / 4, np.pi / 4, 2001)

errores_pi = []
errores_pi4 = []

for n in grados_impares:
    err1 = np.max(np.abs(np.sin(x_pi) - taylor_sin(x_pi, n)))
    err2 = np.max(np.abs(np.sin(x_pi4) - taylor_sin(x_pi4, n)))
    errores_pi.append(err1)
    errores_pi4.append(err2)

plt.figure(figsize=(9, 5.5))
plt.semilogy(grados_impares, errores_pi, 'o-', linewidth=2, markersize=6, label=r'Intervalo $[-\pi, \pi]$')
plt.semilogy(grados_impares, errores_pi4, 's--', linewidth=2, markersize=6, label=r'Intervalo $[-\pi/4, \pi/4]$')
plt.axhline(1e-6, color='red', linestyle=':', linewidth=1.5, label=r'Tolerancia $10^{-6}$')

plt.xticks(grados_impares)
plt.title(r'Error máximo $\max|\sin(x) - P_n(x)|$ frente al grado del polinomio Alejandro López', fontsize=12)
plt.xlabel('Grado del polinomio ($n$)', fontsize=11)
plt.ylabel('Error máximo (escala logarítmica)', fontsize=11)
plt.legend(loc='upper right', frameon=True)
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('grafica_2_error_maximo.png', dpi=300)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def serie_geometrica(x, N):
    x = np.asarray(x, dtype=float)
    termino = np.ones_like(x)
    suma = termino.copy()
    
    for _ in range(N):
        termino = -termino * (x**2)
        suma = suma + termino
        
    return suma

x = np.linspace(-1.5, 1.5, 1000)
f_exacta = 1 / (1 + x**2)

plt.figure(figsize=(9, 5.5))
plt.plot(x, f_exacta, 'k-', linewidth=2.5, label=r'$f(x) = \frac{1}{1+x^2}$')

valores_N = [2, 5, 10, 20]
estilos = ['--', '-.', ':', '--']

for N, estilo in zip(valores_N, estilos):
    plt.plot(x, serie_geometrica(x, N), linestyle=estilo, linewidth=1.8, label=f'$S_{{{N}}}(x)$')

plt.axvline(-1, color='gray', linestyle=':', alpha=0.7)
plt.axvline(1, color='gray', linestyle=':', alpha=0.7)

plt.xlim(-1.5, 1.5)
plt.ylim(-0.5, 1.5)
plt.title(r'Aproximación de $\frac{1}{1+x^2}$ con sumas parciales $S_N(x)$ en $[-1.5, 1.5]$ Alejandro López', fontsize=12)
plt.xlabel('$x$', fontsize=11)
plt.ylabel('$y$', fontsize=11)
plt.legend(loc='lower center', frameon=True, ncol=3)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('grafica_3_serie_geometrica.png', dpi=300)
plt.show()
```
