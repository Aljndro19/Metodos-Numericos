```python
# Ejercicio 1: Método de Newton-Raphson (Algoritmo 2.3)

def newton_raphson(f, df, p0, TOL=1e-6, N0=50):
    """Algoritmo 2.3. Devuelve (p, historia), con historia = [p0, p1, ..., p]."""
    historia = [p0]
    i = 1
    while i <= N0:
        d = df(p0)
        if d == 0:
            raise ZeroDivisionError(f"f'(p0) = 0 en p0 = {p0}: la tangente es horizontal")
        p = p0 - f(p0) / d
        historia.append(p)
        if abs(p - p0) < TOL:
            return p, historia
        i = i + 1
        p0 = p
    raise RuntimeError(f"El método fracasó después de {N0} iteraciones")
```

```python
# Ejercicio 2: Método de la Secante (Algoritmo 2.4)

def secante(f, p0, p1, TOL=1e-6, N0=50):
    """Algoritmo 2.4. Devuelve (p, historia), con historia = [p0, p1, p2, ..., p]."""
    i = 2
    q0 = f(p0)
    q1 = f(p1)
    historia = [p0, p1]

    while i <= N0:
        if q1 == q0:
            raise ZeroDivisionError(f"q1 == q0 = {q1}: la recta secante es horizontal")

        p = p1 - q1 * (p1 - p0) / (q1 - q0)
        historia.append(p)

        if abs(p - p1) < TOL:
            return p, historia

        i = i + 1
        p0, q0 = p1, q1
        p1, q1 = p, f(p)

    raise RuntimeError(f"El método fracasó después de {N0} iteraciones")
```

```python
# Ejercicio 3: Tasa de interés de un préstamo

def saldo_final(i, monto=20000.0, cuota=450.0, meses=60):
    """Saldo después de pagar todas las cuotas, con interés mensual i (0.01 es 1 %)."""
    saldo = monto
    for mes in range(meses):
        saldo = saldo * (1 + i) - cuota
    return saldo


if __name__ == "__main__":
    # --- Pruebas Ejercicio 1 ---
    f1 = lambda x: x**3 + 4*x**2 - 10
    df1 = lambda x: 3*x**2 + 8*x
    p_newton, hist_newton = newton_raphson(f1, df1, p0=1.5, TOL=1e-6)
    print(f"Ejercicio 1 -> p = {p_newton:.6f}, historia = {hist_newton}")

    # --- Pruebas Ejercicio 2 ---
    llamadas = []
    def f2(x):
        llamadas.append(x)
        return x**3 + 4*x**2 - 10

    llamadas.clear()
    p_sec, hist_sec = secante(f2, 1.0, 2.0, TOL=1e-6)
    print(f"Ejercicio 2 -> p = {p_sec:.6f}, iteraciones = {len(hist_sec) - 2}, evaluaciones = {len(llamadas)}")

    # --- Pruebas Ejercicio 3 ---
    print(f"Ejercicio 3 -> saldo_final(0) = {saldo_final(0)}")
    i_mensual, _ = secante(saldo_final, 0.005, 0.02, TOL=1e-10)
    tea = (1 + i_mensual)**12 - 1
    print(f"Ejercicio 3 -> i mensual = {i_mensual * 100:.4f} %, TEA = {tea * 100:.2f} %")
```
